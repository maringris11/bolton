```python
import pandas as pd

def vacio(s):
    return s.isna() | s.astype(str).str.strip().eq("")

df_bolton["tiene_fecha"]    = ~vacio(df_bolton["FECHA"])
df_bolton["tiene_pesos"]    = ~vacio(df_bolton["PESOS"])
df_bolton["tiene_forecast"] = ~vacio(df_bolton["DEMAND FORECAST"]) | ~vacio(df_bolton["PRESUPUESTO"])

print(df_bolton.groupby(["tiene_fecha", "tiene_pesos", "tiene_forecast"]).size())
```

```python
ventas = df_bolton[df_bolton["tiene_pesos"] & ~df_bolton["FECHA"].astype(str).str.strip().eq("INV")].copy()
ventas["PESOS_n"] = pd.to_numeric(ventas["PESOS"], errors="coerce")
ventas["CAJAS_n"] = pd.to_numeric(ventas["CAJAS"], errors="coerce")
ventas["FECHA_d"] = pd.to_datetime(ventas["FECHA"], errors="coerce")
print("ventas:", len(ventas))

# 1. Sin identificación de producto
cols_prod = ["Producto", "Referencia", "Familia", "Categoria"]
sin_prod = ventas[ventas[cols_prod].isna().any(axis=1)]
print("1. sin producto:", len(sin_prod), "| pesos:", sin_prod["PESOS_n"].sum())

# 2. Montos raros
print("2. PESOS no numérico:", ventas["PESOS_n"].isna().sum())
print("   PESOS <= 0:", (ventas["PESOS_n"] <= 0).sum())
print("   CAJAS nulas/<=0:", (ventas["CAJAS_n"].isna() | (ventas["CAJAS_n"] <= 0)).sum())

# 3. Duplicados exactos, ANTES de borrar Factura
print("3. duplicados exactos:", ventas.duplicated(keep=False).sum())

# 4. Fecha vs año/mes
print("4. FECHA inválida:", ventas["FECHA_d"].isna().sum())
bad = ventas[(ventas["FECHA_d"].dt.year != pd.to_numeric(ventas["AÑO"], errors="coerce")) |
             (ventas["FECHA_d"].dt.month != pd.to_numeric(ventas["MES"], errors="coerce"))]
print("   no coincide con AÑO/MES:", len(bad))
print("   rango:", ventas["FECHA_d"].min(), "->", ventas["FECHA_d"].max())

# 5. Sin cliente o canal
print("5. sin cliente:", ventas["Nombre Cliente"].isna().sum(),
      "| sin canal:", ventas["Canal"].isna().sum())
```

```python
# a) PESOS <= 0: ¿ceros o negativos? ¿Qué dice DOCUMENTO?
p = ventas["PESOS_n"]
print("PESOS == 0:", (p == 0).sum(), "| PESOS < 0:", (p < 0).sum(),
      "| suma negativos:", p[p < 0].sum())
print(pd.crosstab(ventas["DOCUMENTO"].fillna("SIN DOC"),
                  p.gt(0).map({True: "PESOS>0", False: "PESOS<=0"})))

# b) FECHA inválida: ¿qué texto traen?
malas = ventas[ventas["FECHA_d"].isna()]
print(malas["FECHA"].astype(str).value_counts().head(10))
print("con PESOS<=0:", (malas["PESOS_n"] <= 0).sum(),
      "| con Semana Año nula:", malas["Semana Año"].isna().sum())

# c) AÑO / MES: ¿qué contienen realmente?
print(ventas["AÑO"].value_counts().head())
print(ventas["MES"].value_counts().head(15))

# d) Duplicados: ¿cuántos sobran y cuántos tienen Factura?
print("filas que sobran (keep='first'):", ventas.duplicated().sum())
print(ventas.loc[ventas.duplicated(keep=False), "Factura"].isna().value_counts())
print("suma PESOS de las que sobran:", ventas.loc[ventas.duplicated(), "PESOS_n"].sum())
```

```python
p, c = ventas["PESOS_n"], ventas["CAJAS_n"]

# a) Escala: ventas brutas, devoluciones y neto
print("positivos:", p[p > 0].sum(), "| negativos:", p[p < 0].sum(), "| neto:", p.sum())

# b) Ceros: ¿tienen cajas?
z = ventas[p == 0]
print("ceros con CAJAS>0:", (z["CAJAS_n"] > 0).sum(), "| sin cajas:", z["CAJAS_n"].fillna(0).eq(0).sum())

# c) FACTURA (-) con monto positivo
raras = ventas[(ventas["DOCUMENTO"] == "FACTURA (-)") & (p > 0)]
print("(-) con PESOS>0:", len(raras), "| suma:", raras["PESOS_n"].sum())

# d) Las fechas 00/01/2020: ¿qué AÑO y MES traen?
f = ventas[ventas["FECHA"].astype(str).str.strip() == "00/01/2020"]
print(f.groupby(["AÑO", "MES"]).size())

p = ventas["PESOS_n"]
cols = [x for x in ventas.columns if x not in ["PESOS_n", "CAJAS_n", "FECHA_d"]]
v = ventas[cols].astype(str)          # todo a texto: evita el choque datetime / time / str

sobran = v.duplicated()
print("sobran con PESOS==0:", (sobran & (p == 0)).sum(), "| con PESOS!=0:", (sobran & (p != 0)).sum())
print("suma PESOS sobrantes != 0:", p[sobran & (p != 0)].sum())

# tamaño de los grupos repetidos
tam = v.groupby(cols, sort=False).size()
print(tam.value_counts().sort_index().head(8))

# muestra de repetidos
rep = ventas[v.duplicated(keep=False)]
print(rep.sort_values(["Factura", "Producto"], key=lambda s: s.astype(str))
         .head(12)[["FECHA", "Factura", "Nombre Cliente", "Producto", "CAJAS", "PESOS"]])
```

```python
import unicodedata

def norm(s):
    return (s.astype(str).str.upper().str.strip()
             .map(lambda x: unicodedata.normalize("NFKD", x).encode("ascii", "ignore").decode()))

p, c = ventas["PESOS_n"], ventas["CAJAS_n"]
cli = norm(ventas["Nombre Cliente"])

interna = cli.str.contains("COLOMBOESPANOLA", na=False)
devol   = (ventas["DOCUMENTO"] == "FACTURA (-)") | (p < 0)

print(cli[interna].value_counts())                       # variantes del nombre
print("interna:", interna.sum(), "| pesos:", p[interna].sum(),
      "| con PESOS>0:", (interna & (p > 0)).sum())
print("devolución:", devol.sum(), "| pesos:", p[devol].sum())
print("ambas:", (interna & devol).sum())

resto = ventas[~interna & ~devol]
r = resto["PESOS_n"]
print("quedan:", len(resto), "| pesos:", r.sum())
print("PESOS==0:", (r == 0).sum(),
      "| de esos sin cajas:", ((r == 0) & (resto["CAJAS_n"].isna() | (resto["CAJAS_n"] <= 0))).sum())

# ¿está 2026 completo?
print(resto[resto["AÑO"].astype(str) == "2026"]["MES"].value_counts())
```

```python
final = resto[resto["PESOS_n"] > 0]
print("ventas finales:", len(final), "| pesos:", final["PESOS_n"].sum())
print("PESOS>0 sin cajas válidas:", (final["CAJAS_n"].isna() | (final["CAJAS_n"] <= 0)).sum())

# a) Las 10 filas internas con monto
print(ventas[interna & (p > 0)][["AÑO", "MES", "Nombre Cliente", "Canal", "Producto", "CAJAS", "PESOS"]])

# b) Duplicados en lo que queda, todavía con Factura
cols = [x for x in final.columns if x not in ["PESOS_n", "CAJAS_n", "FECHA_d"]]
v = final[cols].astype(str)
dup = v.duplicated(keep=False)
print("filas en grupos repetidos:", dup.sum(), "| sobrantes:", v.duplicated().sum(),
      "| pesos sobrantes:", final.loc[v.duplicated(), "PESOS_n"].sum())

fac = final.loc[dup & final["Factura"].notna(), "Factura"].unique()
sub = final[final["Factura"].isin(fac)].assign(dup=dup)
print(sub.groupby("Factura")["dup"].mean().describe())   # cercano a 1 = factura entera repetida

print(final[dup].sample(12, random_state=1)[["Factura", "Nombre Cliente", "Producto", "CAJAS", "PESOS"]])
```
